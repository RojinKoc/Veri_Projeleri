```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
```

```python
data = load_breast_cancer()
X = data.data.T 
y = data.target.reshape(1, X.shape[1])

X_train, X_test, y_train, y_test = train_test_split(X.T, y.T, test_size=0.2, random_state=42)
X_train, X_test, y_train, y_test = X_train.T, X_test.T, y_train.T, y_test.T
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
# Forward & Backward Propagation
def propagate(w, b, X, Y):
    m = X.shape[1] # Örnek sayısı
    
    # Forward
    A = sigmoid(np.dot(w.T, X) + b)
    # Backward
    dw = (1/m) * np.dot(X, (A-Y).T)
    db = (1/m) * np.sum(A-Y)
    
    return dw, db
```

```python
def train_model(X, Y, num_iterations, learning_rate):
    # Initially, we set the weights (w) and the bias (b) to zero
    w = np.zeros((X.shape[0], 1))
    b = 0.0
    
    for i in range(num_iterations):
        dw, db = propagate(w, b, X, Y)
        # Update Parameters
        w = w - learning_rate * dw
        b = b - learning_rate * db
        
    return w, b
```

```python
# Prediction
def predict(w, b, X):
    A = sigmoid(np.dot(w.T, X) + b)
    return (A > 0.5).astype(int) # If greater than 50%, 1 (benign); if less, 0 (malignant)
```

```python
w, b = train_model(X_train, y_train, num_iterations=2000, learning_rate=0.0001)
tahminler = predict(w, b, X_test)
basari = np.mean(tahminler == y_test) * 100

print(f"Model Accuracy Rate Without Libraries: %{basari:.2f}")
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=10000)
model.fit(X_train.T, y_train.T.ravel())

sk_basari = model.score(X_test.T, y_test.T.ravel()) * 100
print(f"Scikit-Learn Model Accuracy Rate: %{sk_basari:.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

tahminler_sk = model.predict(X_test.T)

# Not: y_test ve tahminler (1, n) boyutunda matrisler olduğu için .ravel() ile düz listeye çeviriyoruz
cm_custom = confusion_matrix(y_test.ravel(), tahminler.ravel())
cm_sklearn = confusion_matrix(y_test.ravel(), tahminler_sk.ravel())

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(cm_custom, annot=True, fmt='d', cmap='Blues', ax=axes[0], 
            xticklabels=['Kötü Huylu (0)', 'İyi Huylu (1)'], 
            yticklabels=['Kötü Huylu (0)', 'İyi Huylu (1)'])
axes[0].set_title('Kütüphanesiz Model (Sıfırdan)')
axes[0].set_xlabel('Modelin Tahmini')
axes[0].set_ylabel('Gerçek Durum')

sns.heatmap(cm_sklearn, annot=True, fmt='d', cmap='Greens', ax=axes[1],
            xticklabels=['Kötü Huylu (0)', 'İyi Huylu (1)'], 
            yticklabels=['Kötü Huylu (0)', 'İyi Huylu (1)'])
axes[1].set_title('Scikit-Learn Modeli')
axes[1].set_xlabel('Modelin Tahmini')
axes[1].set_ylabel('Gerçek Durum')

plt.tight_layout()
plt.show()
```
